# now using pipeline

```python
from sklearn.compose import ColumnTransformer
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import OneHotEncoder
from sklearn.tree import DecisionTreeClassifier
from sklearn.pipeline import Pipeline
from sklearn.feature_selection import SelectKBest, chi2
from sklearn.feature_selection import SelectKBest, chi2 
```

```python
df = pd.read_csv('Titanic-Dataset.csv')
```

```python
df.head()
```

# plan
- missing values impute 
- onehot encoding
- scalling
- feature selection
- model traing using decision tree

```python
df.drop(columns=['PassengerId', 'Name', 'Ticket', 'Cabin'], inplace=True)
```

```python
x_train, x_test, y_train, y_test = train_test_split(
    df.drop(columns=['Survived']),
    df['Survived'],
    test_size=0.2,
    random_state=42
)
```

```python
x_train
```

```python
y_train
```

1 step of pipeline

```python
from sklearn.compose import ColumnTransformer

trf1 = ColumnTransformer([
    ('imputer_age', SimpleImputer(), [2]), # 2 is the age index
    ('imputer_embarked', SimpleImputer(strategy='most_frequent'), [6]) # 6th one is embarked index
], remainder='passthrough')
```

2 onehotencoding

```python
trf2 = ColumnTransformer([
    ('onehot_sex_embarked', OneHotEncoder(sparse_output=False, handle_unknown='ignore'), [1,6]),
], remainder='passthrough') # passthough means other columns remain unchanged
```

scallling

```python
trf3 = ColumnTransformer([
    ('scaler', MinMaxScaler(), slice(0,8)), # scaling all columns except onehot encoded columns
])
```

4 feature selection

```python
trf4 = SelectKBest(score_func=chi2, k=5) # select top 5 features
```

5 decision tree

```python
trf5 = DecisionTreeClassifier()
```

# CREATING PIPELINE

```python
pipe = Pipeline([
    ('imputation', trf1),
    ('encoding', trf2),
    ('scaling', trf3),
    ('feature_selection', trf4),
    ('classification', trf5)
])
```

```python
#train the pipeline
pipe.fit(x_train, y_train)
```

```python
pipe.named_steps
```

```python
# code here visualization backtrak debugging
pipe.named_steps['imputation'].transformers_[1][1].statistics_
```

```python
from sklearn import set_config
set_config(display='diagram')
```

```python
y_pred = pipe.predict(x_test)
```

```python
from sklearn.metrics import accuracy_score
accuracy = accuracy_score(y_test, y_pred)
```

```python
print("Accuracy:", accuracy)
```

```python
from sklearn.model_selection import cross_val_score
cross_val_score(pipe, x_train, y_train, cv=5, scoring='accuracy').mean()
```
